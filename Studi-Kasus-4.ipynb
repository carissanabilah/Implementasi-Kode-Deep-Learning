```python
import os
import time
import warnings
from pathlib import Path

import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.base import clone
from sklearn.model_selection import (
    train_test_split,
    StratifiedKFold,
    cross_validate
)
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    RocCurveDisplay
)
from sklearn.inspection import permutation_importance
from IPython.display import display

warnings.filterwarnings("ignore", category=FutureWarning)

# ================================================================
# 0. KONFIGURASI
# ================================================================

SEED = 42
np.random.seed(SEED)

OUTPUT = Path("hasil_ml_heart_disease")
OUTPUT.mkdir(parents=True, exist_ok=True)

URL = (
    "https://archive.ics.uci.edu/ml/machine-learning-databases/"
    "heart-disease/processed.cleveland.data"
)

COLUMNS = [
    "age", "sex", "cp", "trestbps", "chol",
    "fbs", "restecg", "thalach", "exang",
    "oldpeak", "slope", "ca", "thal", "target"
]

NUMERIC = [
    "age", "trestbps", "chol",
    "thalach", "oldpeak", "ca"
]

CATEGORICAL = [
    "sex", "cp", "fbs", "restecg",
    "exang", "slope", "thal"
]

print("=" * 70)
print("IMPLEMENTASI MACHINE LEARNING - HEART DISEASE")
print("=" * 70)

# ================================================================
# 1. PENGUMPULAN DATA NYATA
# ================================================================

print("\n[1] PENGUMPULAN DATA")

try:
    df = pd.read_csv(
        URL,
        header=None,
        names=COLUMNS,
        na_values=["?"]
    )
except Exception as e:
    raise RuntimeError(
        "Dataset gagal diunduh. Periksa koneksi internet "
        "atau ketersediaan server UCI."
    ) from e

for col in COLUMNS:
    df[col] = pd.to_numeric(df[col], errors="coerce")

print("Sumber:", URL)
print("Ukuran awal:", df.shape)

display(df.head())

# ================================================================
# 2. DATA UNDERSTANDING
# ================================================================

print("\n[2] DATA UNDERSTANDING")

print("\nMissing values:")
print(df.isna().sum())

print("\nDuplikasi:", df.duplicated().sum())

df = df.drop_duplicates().reset_index(drop=True)
df = df.dropna(subset=["target"]).copy()

df["target"] = (df["target"] > 0).astype(int)

print("\nDistribusi target:")
print(df["target"].value_counts().sort_index())

print("\nStatistik deskriptif:")
display(df.describe().round(2))

df.to_csv(
    OUTPUT / "dataset_bersih.csv",
    index=False
)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

counts = df["target"].value_counts().sort_index()

axes[0].bar(
    ["Tidak ada penyakit", "Ada penyakit"],
    counts.values,
    color=["steelblue", "coral"]
)
axes[0].set_title("Distribusi Kelas")
axes[0].set_ylabel("Jumlah Pasien")

for i, value in enumerate(counts.values):
    axes[0].text(
        i, value + 2, str(value),
        ha="center"
    )

axes[1].hist(
    [
        df.loc[df.target == 0, "age"],
        df.loc[df.target == 1, "age"]
    ],
    bins=12,
    label=["Tidak ada penyakit", "Ada penyakit"],
    alpha=0.75
)
axes[1].set_title("Distribusi Usia Pasien")
axes[1].set_xlabel("Usia")
axes[1].set_ylabel("Frekuensi")
axes[1].legend()

plt.tight_layout()
plt.savefig(
    OUTPUT / "eksplorasi_dataset.png",
    dpi=150
)
plt.show()

# ================================================================
# 3. PEMBAGIAN DATA
# ================================================================

print("\n[3] PEMBAGIAN DATA")

X = df.drop(columns=["target"])
y = df["target"]


X_dev, X_test, y_dev, y_test = train_test_split(
    X, y,
    test_size=0.20,
    stratify=y,
    random_state=SEED
)

print("Development:", len(X_dev))
print("Testing    :", len(X_test))

print("\nDistribusi kelas:")
print("Development:")
print(y_dev.value_counts(normalize=True).sort_index())
print("Testing:")
print(y_test.value_counts(normalize=True).sort_index())

# ================================================================
# 4. PREPROCESSING TANPA DATA LEAKAGE
# ================================================================

print("\n[4] PREPROCESSING")

numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(
        handle_unknown="ignore"
    ))
])

preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, NUMERIC),
    ("categorical", categorical_pipeline, CATEGORICAL)
])

print("Numeric : median imputation + standardization")
print("Category: mode imputation + one-hot encoding")
print("Seluruh preprocessing berada dalam Pipeline.")

# ================================================================
# 5. DEFINISI MODEL
# ================================================================

print("\n[5] DEFINISI MODEL")

models = {
    "Logistic Regression": Pipeline([
        ("preprocessing", clone(preprocessor)),
        ("classifier", LogisticRegression(
            max_iter=2000,
            random_state=SEED
        ))
    ]),

    "Random Forest": Pipeline([
        ("preprocessing", clone(preprocessor)),
        ("classifier", RandomForestClassifier(
            n_estimators=100,
            max_depth=5,
            min_samples_leaf=2,
            random_state=SEED,
            n_jobs=1
        ))
    ])
}

# ================================================================
# 6. STRATIFIED CROSS-VALIDATION
# ================================================================

print("\n[6] STRATIFIED CROSS-VALIDATION")

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=SEED
)

scoring = {
    "accuracy": "accuracy",
    "precision": "precision",
    "recall": "recall",
    "f1": "f1",
    "roc_auc": "roc_auc"
}

cv_results = []

for name, model in models.items():

    print("\nEvaluasi:", name)
    start = time.perf_counter()

    scores = cross_validate(
        model,
        X_dev,
        y_dev,
        cv=cv,
        scoring=scoring,
        return_train_score=True,
        n_jobs=1,
        error_score="raise"
    )

    duration = time.perf_counter() - start

    row = {
        "Model": name,
        "Accuracy Mean": scores["test_accuracy"].mean(),
        "Precision Mean": scores["test_precision"].mean(),
        "Recall Mean": scores["test_recall"].mean(),
        "F1 Mean": scores["test_f1"].mean(),
        "F1 Std": scores["test_f1"].std(),
        "ROC-AUC Mean": scores["test_roc_auc"].mean(),
        "Train F1 Mean": scores["train_f1"].mean(),
        "CV Time (s)": duration
    }

    cv_results.append(row)

cv_df = pd.DataFrame(cv_results)

print("\nHASIL CROSS-VALIDATION")
display(cv_df.round(4))

cv_df.to_csv(
    OUTPUT / "cross_validation.csv",
    index=False
)

# ================================================================
# 7. PEMILIHAN MODEL
# ================================================================

print("\n[7] PEMILIHAN MODEL")

ranking = cv_df.sort_values(
    ["F1 Mean", "ROC-AUC Mean"],
    ascending=[False, False]
)

best_name = ranking.iloc[0]["Model"]

print("Model terpilih:", best_name)

best_model = clone(models[best_name])

start = time.perf_counter()
best_model.fit(X_dev, y_dev)
fit_time = time.perf_counter() - start

print(f"Waktu pelatihan akhir: {fit_time:.3f} detik")

# ================================================================
# 8. EVALUASI FINAL TEST SET
# ================================================================

print("\n[8] EVALUASI FINAL")

y_pred = best_model.predict(X_test)
y_prob = best_model.predict_proba(X_test)[:, 1]

metrics = {
    "Accuracy": accuracy_score(y_test, y_pred),
    "Precision": precision_score(
        y_test, y_pred, zero_division=0
    ),
    "Recall": recall_score(
        y_test, y_pred, zero_division=0
    ),
    "F1-Score": f1_score(
        y_test, y_pred, zero_division=0
    ),
    "ROC-AUC": roc_auc_score(y_test, y_prob)
}

print("Model:", best_name)

for metric, value in metrics.items():
    print(f"{metric:12s}: {value:.4f}")

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred,
    target_names=["Tidak ada penyakit", "Ada penyakit"],
    zero_division=0
))

final_df = pd.DataFrame([{
    "Model": best_name,
    **metrics
}])

final_df.to_csv(
    OUTPUT / "evaluasi_final.csv",
    index=False
)

# ================================================================
# 9. VISUALISASI PERBANDINGAN MODEL
# ================================================================

print("\n[9] VISUALISASI PERBANDINGAN")

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

x = np.arange(len(cv_df))
width = 0.35

axes[0].bar(
    x - width / 2,
    cv_df["Accuracy Mean"],
    width,
    label="Accuracy"
)
axes[0].bar(
    x + width / 2,
    cv_df["F1 Mean"],
    width,
    label="F1"
)

axes[0].set_xticks(x)
axes[0].set_xticklabels(
    cv_df["Model"],
    rotation=10
)
axes[0].set_ylim(0, 1.05)
axes[0].set_ylabel("Skor")
axes[0].set_title("Perbandingan Cross-Validation")
axes[0].legend()

axes[1].bar(
    x - width / 2,
    cv_df["Train F1 Mean"],
    width,
    label="Training F1"
)
axes[1].bar(
    x + width / 2,
    cv_df["F1 Mean"],
    width,
    label="Validation F1"
)

axes[1].set_xticks(x)
axes[1].set_xticklabels(
    cv_df["Model"],
    rotation=10
)
axes[1].set_ylim(0, 1.05)
axes[1].set_ylabel("F1-Score")
axes[1].set_title("Analisis Generalisasi")
axes[1].legend()

plt.tight_layout()
plt.savefig(
    OUTPUT / "perbandingan_model.png",
    dpi=150
)
plt.show()

print("\nSelisih Training - Validation F1:")

for _, row in cv_df.iterrows():
    gap = row["Train F1 Mean"] - row["F1 Mean"]
    print(f"{row['Model']}: {gap:.4f}")

# ================================================================
# 10. CONFUSION MATRIX DAN ROC CURVE
# ================================================================

print("\n[10] CONFUSION MATRIX DAN ROC")

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

cm = confusion_matrix(
    y_test,
    y_pred,
    labels=[0, 1]
)

ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Tidak ada", "Ada"]
).plot(
    ax=axes[0],
    cmap="Blues",
    colorbar=False
)

axes[0].set_title("Confusion Matrix - Test Set")

RocCurveDisplay.from_predictions(
    y_test,
    y_prob,
    ax=axes[1],
    name=best_name
)

axes[1].plot(
    [0, 1], [0, 1],
    "k--",
    label="Random classifier"
)
axes[1].set_title("ROC Curve - Test Set")
axes[1].legend()

plt.tight_layout()
plt.savefig(
    OUTPUT / "confusion_matrix_roc.png",
    dpi=150
)
plt.show()

tn, fp, fn, tp = cm.ravel()

print("True Negative :", tn)
print("False Positive:", fp)
print("False Negative:", fn)
print("True Positive :", tp)

# ================================================================
# 11. EXPLAINABLE MACHINE LEARNING
# ================================================================

print("\n[11] FEATURE IMPORTANCE")


importance_per_fold = []

for train_idx, val_idx in cv.split(X_dev, y_dev):

    X_fold_train = X_dev.iloc[train_idx]
    X_fold_val = X_dev.iloc[val_idx]
    y_fold_train = y_dev.iloc[train_idx]
    y_fold_val = y_dev.iloc[val_idx]

    fold_model = clone(models[best_name])
    fold_model.fit(X_fold_train, y_fold_train)

    result = permutation_importance(
        fold_model,
        X_fold_val,
        y_fold_val,
        scoring="f1",
        n_repeats=5,
        random_state=SEED,
        n_jobs=1
    )

    importance_per_fold.append(
        result.importances_mean
    )

importance_array = np.array(importance_per_fold)

importance_df = pd.DataFrame({
    "Feature": X_dev.columns,
    "Importance Mean": importance_array.mean(axis=0),
    "Importance Std": importance_array.std(axis=0)
}).sort_values(
    "Importance Mean",
    ascending=False
)

display(importance_df.round(4))

importance_df.to_csv(
    OUTPUT / "feature_importance.csv",
    index=False
)

plot_data = importance_df.iloc[::-1]

plt.figure(figsize=(9, 6))

plt.barh(
    plot_data["Feature"],
    plot_data["Importance Mean"],
    xerr=plot_data["Importance Std"],
    color="steelblue"
)

plt.xlabel("Mean Decrease in F1")
plt.title("Cross-Validated Permutation Importance")
plt.tight_layout()

plt.savefig(
    OUTPUT / "feature_importance.png",
    dpi=150
)
plt.show()

# ================================================================
# 12. PREDIKSI DATA PASIEN ASLI
# ================================================================

print("\n[12] PREDIKSI PASIEN")


sample = X_test.iloc[[0]].copy()
actual = int(y_test.iloc[0])

prediction = int(best_model.predict(sample)[0])
probability = float(
    best_model.predict_proba(sample)[0, 1]
)

print("\nData pemeriksaan pasien:")
display(sample.T)

print("Label aktual:", actual)
print("Hasil prediksi:", prediction)
print(f"Skor probabilitas positif: {probability:.2%}")

if actual == prediction:
    print("Prediksi sesuai label dataset.")
else:
    print("Prediksi tidak sesuai label dataset.")

print(
    "\nPeringatan: model ini hanya untuk pembelajaran, "
    "bukan alat diagnosis medis."
)

# ================================================================
# 13. SIMPAN MODEL DAN HASIL
# ================================================================

print("\n[13] PENYIMPANAN")

model_path = OUTPUT / "model_heart_disease.joblib"

joblib.dump(best_model, model_path)


reloaded_model = joblib.load(model_path)

assert np.array_equal(
    reloaded_model.predict(X_test),
    y_pred
)

prediction_df = X_test.copy()
prediction_df["Actual"] = y_test.to_numpy()
prediction_df["Predicted"] = y_pred
prediction_df["Probability"] = y_prob

prediction_df.to_csv(
    OUTPUT / "prediksi_test.csv",
    index=False
)

print("\nDaftar file hasil:")

for path in sorted(OUTPUT.iterdir()):
    print("-", path.name)

print("\n" + "=" * 70)
print("EKSPERIMEN MACHINE LEARNING SELESAI")
print("=" * 70)
```
